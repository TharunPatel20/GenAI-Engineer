## Step 3: check the skeleton

`app_skeleton/` holds one file for each part of the app:

- `config.py`: settings, model names, and folder paths (M2, M3, M4)
- `ingest.py`: load, chunk, embed, and save the index (M2, from project 05)
- `retrieval.py`: keyword, vector, and hybrid search (M3, from project 06)
- `agent.py`: the LangGraph graph with a search tool (M4, from project 08)
- `safety.py`: input, agent, and output checks (M6)
- `app.py`: the entry point and terminal loop (M5)

Each TODO in these files names the project whose pattern to reuse. The cell below lists the files and runs `app.py`. You should see only the placeholder message. The skeleton modules are not imported here, because they are stubs.

```python
import os
import subprocess
import sys
from pathlib import Path

SKELETON = Path("app_skeleton")

if not SKELETON.exists():
    print("app_skeleton/ not found. Run Step 0 first, so the folder is the current folder.")
else:
    print("Files in app_skeleton/:")
    for root, dirs, files in os.walk(SKELETON):
        dirs[:] = [d for d in dirs if d != "__pycache__"]
        for name in sorted(files):
            print("  ", os.path.join(root, name))

    print()
    print("Running app_skeleton/app.py (placeholder, not the finished app):")
    result = subprocess.run([sys.executable, str(SKELETON / "app.py")], capture_output=True, text=True)
    print(result.stdout.strip())
    if result.returncode != 0:
        print("app.py exited with an error:")
        print(result.stderr.strip())
```
